## Messages
### Role, Content, Metadata

```python
from langchain_ollama import ChatOllama

model_name = ("qwen3-0.6b:latest", "gemma3:1b")[0]
SYSTEM_PROMPT = "You are a helpfull assistant."

model = ChatOllama(model=model_name, temperature=0)
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("You are a programmer"),
    HumanMessage("Write a code that calculate sum of numbers")
]

response = model.invoke(messages)
print(response.content)
```

```python
# Detailed info to the LLM through System message
from langchain.messages import SystemMessage, HumanMessage, AIMessage

sys_msg = SystemMessage("""
You are a senior Python developer with expertise in calculating.
Always provide code example and explain your reasoning.
Be concise but through in your explanations.""")

messages = [
    sys_msg,
    HumanMessage("Write a code that calculate sum of numbers")
]

response = model.invoke(messages)
print(response.content)
```

```python
hmn_msg = HumanMessage(
    content="Hello!",
    name="Alice",
    id="123582"
)

messages = [
    hmn_msg,
]

response = model.invoke(messages)
print(response)
```

```python
from langchain.messages import AIMessage, HumanMessage, SystemMessage

messages = [
    SystemMessage("You are a helpfull assistant."),
    HumanMessage("Can you help me?"),
    AIMessage("I'd be happy to help you with that question!"),
    HumanMessage("Great! What's 2 + 2?")
]

response = model.invoke(messages)
print(response.content)
```

```python
from langchain.messages import AIMessage, ToolMessage, HumanMessage

ai_msg = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "San Francisco"},
        "id": "call_123"
    }]
)

weather_result = "Sunny, 42°C"
tool_msg = ToolMessage(
    content=weather_result,
    tool_call_id="call_123" # Must match the call ID
)

messages = [
    HumanMessage("What's the weather in San Francisco?"),
    ai_msg,
    tool_msg
]

response = model.invoke(messages)
print(response.content)
```
