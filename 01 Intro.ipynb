```python
import os
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_ollama import ChatOllama

load_dotenv()
```

```python
model_name = "qwen3-0.6b:latest"
SYSTEM_PROMPT = "You are a helpfull assistant."
```

```python
model = ChatOllama(model=model_name, temperature=0)
def get_weather(city):
    """Get the weather for a city"""
    return f"The weather of {city} is sunny."
```

```python
agent = create_agent(
    model=model, 
    tools=[get_weather], 
    system_prompt=SYSTEM_PROMPT
    )
```

```python
agent.invoke({"messages": [{"role": "user", "content": "What is the weather of Newyork?"}]})
```

```python
response = agent.invoke({"messages": [{"role": "user", "content": "What is the weather of Newyork?"}]})
```
