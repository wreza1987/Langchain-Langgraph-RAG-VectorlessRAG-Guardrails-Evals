## Tools

```python
import os
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_ollama import ChatOllama

load_dotenv()
```

```python
model_name = "qwen3-0.6b:latest"
SYSTEM_PROMPT = "You are a helpfull assistant."

model = ChatOllama(model=model_name, temperature=0)
```

```python
model.invoke("What is the weather of Newyork?")
```

```python
from langchain.tools import tool

@tool
def get_weather(city: str) -> str:
    """Get the weather at a city"""
    return f"The weather of {city} is sunny."

model_with_tools = model.bind_tools([get_weather])
```

```python
model_with_tools.invoke("What is the weather of Newyork?")
```

```python
response = model_with_tools.invoke("What is the weather like in Boston?")
print(response)

for tc in response.tool_calls:
    print(f"Tool: {tc['name']}")
    print(f"Args: {tc['args']}")
```

### Tool execution

```python
messages = [{"role": "user", "content": "What's the weather in Tehran?"}]
ai_msg = model_with_tools.invoke(messages)

messages.append(ai_msg)

messages
```

```python
for tc in ai_msg.tool_calls:
    tool_result = get_weather.invoke(tc)
    messages.append(tool_result)

messages
```

```python
final_response = model_with_tools.invoke(messages)

final_response.text
```
